```python
PROVINCE = 'Banteay Meanchey'   # e.g. 'Battambang', 'Siem Reap', 'Pursat'
DATE = ''          # e.g. '2026-09-28'. Empty = the newest radar image.
YEARS = 3          # how many previous years to compare with

args = f'--years {YEARS} --province "{PROVINCE}"' + (f' --date {DATE}' if DATE else '')
out = !python scripts/colab_check_flood.py {args}
print(chr(10).join(l for l in out if 'Warning' not in l and 'warnings.warn' not in l))

from IPython.display import Image, Markdown, display
res = [l.split('=', 1)[1] for l in out if l.startswith('RESULT_DIR=')]
if res:
    display(Markdown(open(f'{res[-1]}/report.md', encoding='utf-8').read().replace('![map](map.png)', '')))
    display(Image(f'{res[-1]}/map.png'))
```
